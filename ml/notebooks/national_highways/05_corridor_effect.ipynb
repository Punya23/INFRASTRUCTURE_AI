Built-up growth 2000→2020 (GHSL, % of each 1 km² cell) within 0–2 km vs 2–10 km of expressways that opened between the two epochs, compared with the same rings around corridors not yet open. Descriptive, not causal: corridors are built where growth is expected, and only two epochs are used.
