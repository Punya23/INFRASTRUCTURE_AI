# Geography EDA

## SECTION 1: Project Objective
Build a clean, reproducible, India-wide geographic foundation for the LokDrishti AI Public Transport system.
The geographic hierarchy is:
India ↓ State/UT ↓ District ↓ Sub-District ↓ Urban Local Body (ULB) ↓ Ward


## SECTION 2: Load datasets

```python
import pandas as pd

states = pd.read_csv('../data/processed/states_clean.csv')
districts = pd.read_csv('../data/processed/districts_clean.csv')
subdistricts = pd.read_csv('../data/processed/subdistricts_clean.csv')
ulbs = pd.read_csv('../data/processed/ulbs_clean.csv')
wards = pd.read_csv('../data/processed/wards_all_available_clean.csv')
```

## SECTION 3: Dataset dimensions

```python
print(f"States: {states.shape}")
print(f"Districts: {districts.shape}")
print(f"Subdistricts: {subdistricts.shape}")
print(f"ULBs: {ulbs.shape}")
print(f"Wards: {wards.shape}")
```

## SECTION 4: Column inspection

```python
print(f"States cols: {states.columns.tolist()}")
print(f"Wards cols: {wards.columns.tolist()}")
```

## SECTION 5: Data types

```python
print(wards.dtypes)
```

## SECTION 6: Missing values

```python
print("Missing percentages in Wards:")
print((wards.isna().mean() * 100).round(2).sort_values(ascending=False))
```

## SECTION 7: Duplicate analysis

```python
dup_ward = wards.duplicated(subset=['local_body_code', 'ward_code']).sum()
print(f"Duplicate (local_body_code, ward_code) in Wards: {dup_ward}")
```

## SECTION 8: State analysis

```python
print(f"Total States/UTs: {states['state_code'].nunique()}")
```

## SECTION 9: District analysis

```python
print(f"Total Districts: {districts['district_code'].nunique()}")
```

## SECTION 10: Subdistrict analysis

```python
print(f"Total Subdistricts: {subdistricts['subdistrict_code'].nunique()}")
```

## SECTION 11: ULB analysis

```python
print(f"Total ULBs: {ulbs['local_body_code'].nunique()}")
ulb_per_st = ulbs.groupby('state_name')['local_body_code'].nunique()
print(ulb_per_st.describe())
```

## SECTION 12: Ward analysis

```python
print(f"Total Wards: {wards['ward_code'].nunique()}")
print(wards.groupby('ward_source_state')['ward_code'].nunique())
```

## SECTION 13: Cross-table validation

```python
ward_ulb_codes = set(wards['local_body_code'].dropna())
ulb_codes = set(ulbs['local_body_code'].dropna())
unmatched = ward_ulb_codes - ulb_codes
print(f"Wards local_body_codes unmatched in ULB master: {len(unmatched)}")
```

## SECTION 14: Coverage limitations
Ward coverage is currently limited to Maharashtra, Karnataka, Tamil Nadu, and Delhi.

## SECTION 15: Final data quality summary
Data has been deduplicated, standardized and mostly validated across geographies. The major limitation is Ward coverage (only 4 states available). The pipelines are idempotent and verifiable.
