# TGSRTC Stops EDA
Exploratory Data Analysis of TGSRTC Hyderabad Stops CSV.

```
import pandas as pd
from pathlib import Path

PROCESSED_DIR = Path('../data/processed')
```

## 1. Overview

```
stops = pd.read_csv(PROCESSED_DIR / 'tgsrtc_stops_clean.csv')
print(f'Total stops: {len(stops)}')
print(f'Unique stop IDs: {stops.stop_id.nunique()}')
print('Missing values:')
print(stops.isnull().sum())
print(f'Duplicate coordinates: {stops.duplicated(subset=["stop_lat", "stop_lon"]).sum()}')
```

## 2. Geography

```
print('Latitude bounds:', stops.stop_lat.min(), '-', stops.stop_lat.max())
print('Longitude bounds:', stops.stop_lon.min(), '-', stops.stop_lon.max())
```
