# TGSRTC GTFS EDA
Exploratory Data Analysis of TGSRTC GTFS feed.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROCESSED_DIR = Path('../data/processed')
```

## 1. Stops

```
stops = pd.read_csv(PROCESSED_DIR / 'stops_clean.csv')
print(f'Total stops: {len(stops)}')
print(f'Unique stops: {stops.stop_id.nunique()}')
print('Missing coords:', stops['stop_lat'].isnull().sum())
print('Duplicate coords:', stops.duplicated(subset=['stop_lat', 'stop_lon']).sum())
```

## 2. Routes

```
routes = pd.read_csv(PROCESSED_DIR / 'routes_clean.csv')
print(f'Total routes: {len(routes)}')
print('Route Types:')
print(routes['route_type'].value_counts())
routes['route_type'].value_counts().plot(kind='bar')
plt.title('Routes by Type')
plt.show()
```

## 3. Trips

```
trips = pd.read_csv(PROCESSED_DIR / 'trips_clean.csv')
print(f'Total trips: {len(trips)}')
trips_per_route = trips.groupby('route_id').size()
print('Trips per route (top 5):')
print(trips_per_route.sort_values(ascending=False).head())
trips_per_route.plot(kind='hist', bins=50)
plt.title('Trips per Route')
plt.show()
```

## 4. Stop Times

```
stop_times = pd.read_csv(PROCESSED_DIR / 'stop_times_clean.csv')
print(f'Stop time records: {len(stop_times)}')
stops_per_trip = stop_times.groupby('trip_id').size()
print(f'Avg stops per trip: {stops_per_trip.mean():.1f}')
print(f'Min stops: {stops_per_trip.min()}, Max stops: {stops_per_trip.max()}')
stops_per_trip.plot(kind='hist', bins=50)
plt.title('Stops per Trip')
plt.show()
```

## 5. Calendar

```
calendar = pd.read_csv(PROCESSED_DIR / 'calendar_clean.csv')
print('Service Days:')
print(calendar[['monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday', 'sunday']].sum())
```
