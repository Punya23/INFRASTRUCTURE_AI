# Smart Cities Bus Fleet EDA
Exploratory Data Analysis of Smart Cities Fleet Data.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROCESSED_DIR = Path('../data/processed')
```

## 1. Fleet Overview

```
fleet = pd.read_csv(PROCESSED_DIR / 'smartcities_bus_fleet_clean.csv')
print(f'Total records: {len(fleet)}')
print(f'Cities represented: {fleet["City Name"].nunique()}')
print('Missing values:')
print(fleet.isnull().sum())
```

## 2. Fleet by City

```
city_totals = fleet.groupby('City Name')['No. of buses of that type'].sum().sort_values(ascending=False)
print('Top 10 Cities by Fleet:')
print(city_totals.head(10))
city_totals.head(15).plot(kind='bar', figsize=(10,5))
plt.title('Top 15 Smart Cities by Bus Fleet')
plt.ylabel('Number of Buses')
plt.show()
```

## 3. Bus Types

```
type_totals = fleet.groupby('Type Of Bus (Ac / Non Ac)')['No. of buses of that type'].sum()
print(type_totals)
type_totals.plot(kind='pie', autopct='%1.1f%%')
plt.title('Bus Fleet by Type')
plt.show()
```
